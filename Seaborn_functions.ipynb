```python
#sns.scatterplot():It is used to show the relationship between 2 numerical variables.
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x=[1, 2, 3, 4], y=[10, 20, 15, 30])
plt.show()
```

```python
#sns.lineplot():It is used to show trends or changes in data.
sns.lineplot(x=[1, 2, 3, 4], y=[10, 20, 15, 30])
plt.show()
```

```python
#sns.barplot(): It is used to compare numerical values between different categories.
subjects = ['Python', 'Java', 'SQL', 'ML']
marks = [85, 72, 90, 78]

sns.barplot(x=subjects, y=marks)
plt.show()
```

```python
#sns.countplot(): It is used to count how many times each category occurs.
courses = ['AI', 'CSE', 'AI', 'DS', 'AI', 'CSE']

sns.countplot(x=courses)
plt.show()
```

```python
#sns.histplot():It is used to show the distribution of numerical data.
ages = [18, 19, 19, 20, 20, 20, 21, 22, 22, 23, 24]

sns.histplot(ages, bins=5)
plt.show()
```

```python
#sns.boxplot():It used to identify the spread of data and possible outliers.
salary = [20, 22, 25, 27, 30, 32, 35, 80]

sns.boxplot(y=salary)
plt.show()
```

```python
#sns.violinplot():It used to show both the distribution and spread of data.
sns.violinplot(x=[10, 12, 13, 15, 20, 25])
plt.show()
```

```python
#sns.pairplot():It shows relationships between multiple numerical columns.
import pandas as pd

df = pd.DataFrame({
    'Age': [20, 21, 22, 23],
    'Marks': [60, 70, 80, 90],
    'Salary': [20, 25, 30, 35]
})

sns.pairplot(df)
plt.show()
```

```python
#sns.heatmap(): It is used to represent numerical values using colors, commonly for correlation.
df = pd.DataFrame({
    'Age': [20, 21, 22, 23],
    'Marks': [60, 70, 80, 90]
})

sns.heatmap(df.corr(), annot=True)
plt.show()
```

```python
#sns.regplot(): It isused to show the relationship between two variables with a regression line.
sns.regplot(x=[1, 2, 3, 4], y=[10, 20, 25, 40])
plt.show()

```

```python
#sns.swarmplot():It shows individual data points without overlapping.
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Sales', 'Sales'],
    'Salary': [30, 35, 28, 32, 25, 40]
}

sns.swarmplot(x='Department', y='Salary', data=data)
plt.show()
```

```python
#sns.stripplot():It displays individual observations as points for different categories.
data = {
    'Course': ['Python', 'Python', 'Java', 'Java', 'SQL', 'SQL'],
    'Marks': [80, 90, 70, 85, 75, 95]
}

sns.stripplot(x='Course', y='Marks', data=data)
plt.show()
```

```python
#sns.pointplot():It shows the average value of each category using points.
data = {
    'Subject': ['PowerBI', 'PowerBI', 'Python', 'Python', 'SQL', 'SQL'],
    'Marks': [70, 80, 85, 90, 75, 85]
}

sns.pointplot(x='Subject', y='Marks', data=data)
plt.show()
```

```python
#sns.kdeplot():It shows the probability density or smooth distribution of numerical data.
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85]

sns.kdeplot(marks)
plt.show()
```

```python
#sns.ecdfplot():It shows the cumulative distribution of observations.
marks = [40, 50, 60, 65, 70, 75, 80, 90]

sns.ecdfplot(marks)
plt.show()
```

```python
#sns.jointplot():It shows the relationship between two variables along with their individual distributions.
df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 60, 75, 85]
})

sns.lmplot(x='Hours', y='Marks', data=df)
plt.show()
```

```python
#sns.lmplot():It shows a scatterplot with a linear regression line.
df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 60, 75, 85]
})

sns.lmplot(x='Hours', y='Marks', data=df)
plt.show()
```

```python
#sns.catplot():It creates different types of categorical plots using one function.
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female'],
    'Marks': [70, 85, 80, 90]
})

sns.catplot(x='Gender', y='Marks', data=df, kind='bar')
plt.show()
```

```python
#sns.FacetGrid():It creates multiple graphs based on different categories.
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female'],
    'Marks': [70, 85, 80, 90],
    'Subject': ['Maths', 'Maths', 'Python', 'Python']
})

g = sns.FacetGrid(df, col='Subject')
g.map_dataframe(sns.scatterplot, x='Gender', y='Marks')
plt.show()
```

```python
#sns.set_theme():sets the overall visual style of Seaborn graphs.
sns.set_theme(style='darkgrid')

sns.barplot(x=['A', 'B', 'C'], y=[20, 35, 30])
plt.show()
```

```python
#sns.cubehelix_palette(): It creates a smooth color sequence that is useful for showing increasing or decreasing values.
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_palette(sns.cubehelix_palette(4))

sns.barplot(x=['A', 'B', 'C', 'D'], y=[20, 35, 25, 40])
plt.show()
```

```python
#sns.light_palette():It creates a palette that starts from a light color and becomes darker.
palette = sns.light_palette("blue", 5)

sns.barplot(
    x=['A', 'B', 'C', 'D', 'E'],
    y=[10, 20, 30, 40, 50],
    hue=['A', 'B', 'C', 'D', 'E'],
    palette=palette,
    legend=False
)

plt.show()
```

```python
#sns.dark_palette():It creates a palette that starts from a dark color and becomes lighter.
palette = sns.dark_palette("green", 5)

sns.barplot(
    x=['A', 'B', 'C', 'D', 'E'],
    y=[10, 20, 30, 40, 50],
    hue=['A', 'B', 'C', 'D', 'E'],
    palette=palette,
    legend=False
)

plt.show()
```

```python
#sns.blend_palette():It creates a palette by smoothly blending two or more colors.
palette = sns.blend_palette(["blue", "white", "red"], 5)

sns.barplot(
    x=['A', 'B', 'C', 'D', 'E'],
    y=[10, 20, 30, 40, 50],
    hue=['A', 'B', 'C', 'D', 'E'],
    palette=palette,
    legend=False
)

plt.show()
```

```python
#sns.diverging_palette():It creates a palette with two different colors meeting around a central value.
palette = sns.diverging_palette(240, 10, n=7)

sns.heatmap(
    [[-3, -2, -1, 0, 1, 2, 3]],
    cmap=sns.diverging_palette(240, 10, as_cmap=True),
    annot=True
)

plt.show()
```

```python
#sns.reset_defaults():It resets Seaborn settings back to its default configuration.
sns.set_theme(style="darkgrid")

sns.reset_defaults()

sns.barplot(
    x=['A', 'B', 'C'],
    y=[20, 30, 25]
)

plt.show()
```

```python
#sns.relplot(): Creates scatter or line plots and supports easy faceting into subplots by category.
sns.relplot(x=[1, 2, 3, 4], y=[10, 20, 15, 30], kind='scatter')
plt.show()
```

```python
#sns.displot(): A figure-level function to plot distributions (histogram, KDE, ECDF) with faceting support.
marks = [45, 50, 55, 60, 65, 70, 75, 80]
sns.displot(marks, kind='hist')
plt.show()
```

```python
#sns.desaturate(): It reduces the intensity of a given color.
color = sns.desaturate("red", 0.3)
sns.barplot(x=['A', 'B', 'C'], y=[20, 30, 25], color=color)
plt.show()

```

```python
#sns.set(): A shortcut for set_theme() that applies default Seaborn styling to plots.
sns.set()
sns.barplot(x=['A', 'B', 'C'], y=[20, 30, 25])
plt.show()
```
