```python
#np.poly():Creates a polynomial from its roots
a = np.array([1, 2, 3])
print(np.poly(a))
```

```python
#np.polyval():Evaluates a polynomial at specific values
p = [1, -5, 6]

print(np.polyval(p, 2))
```

```python
#np.polyfit():Fits a polynomial of a specified degree to data
x = np.array([1, 2, 3, 4])
y = np.array([1, 4, 9, 16])

coeff = np.polyfit(x, y, 2)

print(coeff)
```

```python
#np.polyder():Computes the derivative of a polynomial
p = [1, -5, 6]

print(np.polyder(p))
```

```python
#np.polyint():Finds the integral of a polynomial
p = [2, 3]

print(np.polyint(p))
```

```python
#np.polyadd():Adds two polynomials
p1 = [1, 2]
p2 = [3, 4]

print(np.polyadd(p1, p2))
```

```python
#np.polysub():Subtracts two polynomials
p1 = [5, 7]
p2 = [2, 3]

print(np.polysub(p1, p2))
```

```python
#np.polymul():Multiplies two polynomials
p1 = [1, 2]
p2 = [3, 4]
print(np.polymul(p1, p2))
```

```python
#np.polydiv():Divides two polynomials
p1 = [1, 5, 6]
p2 = [1, 2]

q, r = np.polydiv(p1, p2)

print("Quotient:", q)
print("Remainder:", r)
```
