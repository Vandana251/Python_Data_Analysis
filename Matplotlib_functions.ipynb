```python
import matplotlib.pyplot as plt
```

```python
#plt.plot():It draws a line graph.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.show()
```

```python
#plt.scatter():It creates a scatter plot to show relationships between 2 variables.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.scatter(x, y)
plt.show()
```

```python
#plt.scatter():It creates a scatter plot to show relationships between two variables.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.scatter(x, y)
plt.show()
```

```python
#plt.bar():It creates a vertical bar chart.
names = ["A", "B", "C"]
marks = [80, 90, 70]

plt.bar(names, marks)
plt.show()
```

```python
#plt.barh():It creates a horizontal bar chart.
names = ["A", "B", "C"]
marks = [80, 90, 70]

plt.barh(names, marks)
plt.show()
```

```python
#plt.hist(): It shows the distribution/frequency of numerical data.
values = [40, 30, 20, 10]
labels = ["A", "B", "C", "D"]

plt.pie(values, labels=labels)
plt.show()
```

```python
#plt.xlabel():It adds a label to the X-axis.
plt.plot([1, 2, 3], [10, 15, 20])
plt.xlabel("Days")
plt.show()
```

```python
#plt.ylabel():It adds a label to the Y-axis.
plt.plot([1, 2, 3], [10, 20, 30])
plt.ylabel("Sales")
plt.show()
```

```python
#plt.legend():It displays labels for multiple plots.
x = [1, 2, 3]

plt.plot(x, [10, 20, 30], label="Sales")
plt.plot(x, [5, 15, 25], label="Profit")

plt.legend()
plt.show()
```

```python
#plt.grid(): It adds grid lines to the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.grid()
plt.show()
```

```python
#plt.xlim():It sets the X-axis range.
plt.plot([1, 2, 3, 4], [10, 20, 30, 40])
plt.xlim(1, 3)
plt.show()
```

```python
#plt.ylim():It sets the Y-axis range.
plt.plot([1, 2, 3], [10, 20, 30])
plt.ylim(0, 50)
plt.show()
```

```python
#plt.xticks():It changes X-axis tick positions/labels.
plt.plot([1, 2, 3], [10, 20, 30])
plt.xticks([1, 2, 3], ["Jan", "Feb", "Mar"])
plt.show()
```

```python
#plt.figure():It creates a new figure/window for plotting.
plt.figure(figsize=(8, 5))
plt.plot([1, 2, 3], [10, 20, 30])
plt.show()
```

```python
#plt.subplot():It creates multiple graphs in one figure.
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [10, 20, 30])

plt.subplot(1, 2, 2)
plt.bar([1, 2, 3], [10, 20, 30])

plt.show()
```

```python
#plt.show(): It displays the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.show()
```

```python
#plt.savefig(): It saves the graph as an image file.
plt.plot([1, 2, 3], [10, 20, 30])
plt.savefig("graph.png")
plt.show()
```

```python
#plt.close(): It closes the current figure.
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [10, 20, 30])
plt.close()
```

Output:
No graph will be displayed because plt.close() closes the figure immediately.

```python
#plt.fill_between(): It fills the area between two lines.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.fill_between(x, y)
plt.show()
```

```python
#plt.stackplot():It creates a stacked area chart.
x = [1, 2, 3, 4]

a = [10, 20, 15, 25]
b = [5, 10, 20, 15]

plt.stackplot(x, a, b)
plt.show()
```

```python
#plt.step():It creates a step-style graph.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.step(x, y)
plt.show()
```

```python
#plt.stem():It creates a stem plot for discrete values.
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.stem(x, y)
plt.show()
```

```python
#plt.eventplot():It displays positions of events along an axis.
events = [1, 2, 3, 5, 7]

plt.eventplot(events)
plt.show()
```

```python
#plt.contour():It creates contour lines for 3-dimensional data.
import numpy as np

x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contour(X, Y, Z)
plt.show()
```

```python
#plt.title(): It adds a title to the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("Sales Trend")
plt.show()
```

```python
#plt.axhline(): Draws a horizontal reference line across the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.axhline(y=20, color='red')
plt.show()
```

```python
#plt.axvline(): Draws a vertical reference line across the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.axvline(x=2, color='green')
plt.show()
```

```python
#plt.text(): Adds free text at a given position on the graph.
plt.plot([1, 2, 3], [10, 20, 30])
plt.text(2, 25, "Growth Point")
plt.show()
```

```python
#plt.colorbar(): Adds a color scale bar next to a plot to show value ranges (used with heatmaps/scatter with color).
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
colors = [1, 2, 3, 4]

plt.scatter(x, y, c=colors, cmap='viridis')
plt.colorbar()
plt.show()
```

```python
#plt.annotate(): Adds a text label pointing to a specific data point.
plt.plot([1, 2, 3], [10, 20, 30])
plt.annotate("Peak", xy=(3, 30), xytext=(2, 35),
             arrowprops=dict(facecolor='black'))
plt.show()
```

```python
#plt.tight_layout(): Automatically adjusts spacing between subplots so labels/titles don't overlap.
fig, ax = plt.subplots(1, 2)
ax[0].plot([1, 2, 3], [10, 20, 30])
ax[1].bar([1, 2, 3], [10, 20, 30])
plt.tight_layout()
plt.show()
```

```python
#plt.yticks(): Changes Y-axis tick positions/labels.
plt.plot([1, 2, 3], [10, 20, 30])
plt.yticks([10, 20, 30], ["Low", "Medium", "High"])
plt.show()
```

```python
#plt.twinx(): Creates a second Y-axis sharing the same X-axis, useful for comparing two different scales.
fig, ax1 = plt.subplots()
ax1.plot([1, 2, 3], [10, 20, 30], color='blue')

ax2 = ax1.twinx()
ax2.plot([1, 2, 3], [100, 200, 150], color='red')
plt.show()
```

```python
#plt.imshow(): Displays image data or a 2D array as an image/heatmap.
import numpy as np

data = np.random.rand(5, 5)
plt.imshow(data, cmap='hot')
plt.colorbar()
plt.show()
```
