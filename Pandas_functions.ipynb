```python
#pd.Series():stores data in a single column with an index
import pandas as pd

s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
#pd.DataFrame():It is like an Excel table containing rows and columns.
import pandas as pd

df = pd.DataFrame({
    'Name': ['R', 'S'],
    'Marks': [80, 90]
})

print(df)
```

```python
#df.head(): It is used to quickly see the beginning of the dataset.
print(df.head())
```

```python
#df.tail(): It is used to see the ending rows of the dataset.
print(df.tail())
```

```python
#df.shape: It returns the number of rows and columns in the dataframe.
print(df.shape)
```

```python
#df.columns:It returns the names of all columns in the dataset.
print(df.columns)
```

```python
#df.index:It returns the row index information.
print(df.index)
```

```python
#df.info():It gives information about columns, data types, rows, & missing values.
df.info()
```

```python
#describe(): It gives statistics such as count, mean, standard deviation, minimum & maximum.
print(df.describe())
```

```python
#df.dtypes: It shows the data type of each column like integer, float, string, etc.
print(df.dtypes)
```

```python
#df.loc[]:It selects rows and columns using labels.
print(df.loc[0])
```

```python
#df.iloc[]:It selects rows and columns using integer positions.
print(df.iloc[0])
```

```python
#df.isnull():True means the value is missing; False means it is present.
print(df.isnull())
```

```python
#df.dropna():It removes rows where values are missing.
df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Marks': [80, None, 70]
})

print(df.dropna())
```

```python
#df.fillna():It is used to handle missing values instead of deleting rows.
print(df.fillna(0))
```

```python
#df.sort_values():It arranges the data in ascending or descending order.
df = pd.DataFrame({
    'Name': ['Ram', 'Sita', 'Ravi'],
    'Marks': [80, 90, 70]
})

print(df.sort_values('Marks'))
```

```python
#df.drop():It is used to delete unwanted rows or columns.
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ram', 'Sita', 'Ravi'],
    'Marks': [80, 90, 70],
    'Age': [20, 21, 22]
})

print(df.drop('Age', axis=1))
```

```python
#df.rename():It changes the name of rows or columns.
print(df.rename(columns={'Marks': 'Score'}))
```

```python
#df.replace():It replaces one value with another value.
df = pd.DataFrame({
    'Name': ['Ram', 'Sita', 'Ravi'],
    'City': ['Hyderabad', 'Hyderabad', 'Vijayawada']
})

print(df.replace('Hyderabad', 'Vizag'))
```

```python
#df.groupby():Groups data based on a particular column like mean, sum, count.
df = pd.DataFrame({
    'Branch': ['CSE', 'ECE', 'EEE', 'AI'],
    'Marks': [80, 70, 90, 60]
})

print(df.groupby('Branch')['Marks'].mean())
```

```python
#df.mean(): It calculates the average of numerical values.
df = pd.DataFrame({
    'Marks': [80, 90, 70]
})

print(df['Marks'].mean())
```

```python
#df.median():It finds the middle value of numerical data.
print(df['Marks'].median())
```

```python
#df.mode():It finds the most repeated value.
df = pd.DataFrame({
    'Marks': [80, 90, 80, 70]
})

print(df['Marks'].mode())
```

```python
#df.sum():It calculates the total of numerical values.
df = pd.DataFrame({
    'Marks': [80, 90, 70]
})

print(df['Marks'].sum())
```

```python
#df.min(): It returns the smallest value.
print(df['Marks'].min())
```

```python
#df.max(): It returns the largest value.
print(df['Marks'].max())
```

```python
#df.count(): It counts non-missing values in each column.
print(df['Marks'].count())
```

```python
#df.std(): It calculates the standard deviation of numerical data.
print(df['Marks'].std())
```

```python
#df.var():It calculates the variance of numerical data.
print(df['Marks'].var())
```

```python
#df.corr():It calculates the correlation between numerical columns.
df = pd.DataFrame({
    'Hours': [1, 2, 3, 4],
    'Marks': [20, 40, 60, 80]
})

print(df.corr())
```

```python
#df.apply():It applies a function to each value, row, or column.
import pandas as pd

df = pd.DataFrame({
    'Marks': [50, 60, 70]
})

df['Marks'] = df['Marks'].apply(lambda x: x + 5)

print(df)
```

```python

#df.map():It applies a function to every value in a Series.
df = pd.DataFrame({
    'Marks': [50, 60, 70]
})

print(df['Marks'].map(lambda x: x * 2))
```

```python
#df.sort_values():It sorts rows based on one or more columns.
df = pd.DataFrame({
    'Name': ['Ram', 'Sita', 'Ravi'],
    'Marks': [80, 95, 70]
})

print(df.sort_values('Marks', ascending=False))
```

```python
#df.sort_index(): It arranges rows according to their index values.
df = pd.DataFrame({
    'Marks': [80, 90, 70]
}, index=[2, 0, 1])

print(df.sort_index())
```

```python
#pd.concat():It combines two or more DataFrames vertically or horizontally.
df1 = pd.DataFrame({
    'Name': ['Ram', 'Sita']
})

df2 = pd.DataFrame({
    'Name': ['Ravi', 'Anu']
})

print(pd.concat([df1, df2], ignore_index=True))
```

```python
#pd.merge():It combines dataframes using a common column.
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Ram', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [1, 2],
    'Marks': [80, 90]
})

print(pd.merge(df1, df2, on='ID'))
```

```python
#df.join():It joins two DataFrames using their index.
df1 = pd.DataFrame({
    'Name': ['Ram', 'Sita']
}, index=[1, 2])

df2 = pd.DataFrame({
    'Marks': [80, 90]
}, index=[1, 2])

print(df1.join(df2))
```

```python
#df.shift():It moves data by a specified number of rows.
df = pd.DataFrame({
    'Marks': [50, 60, 70]
})

print(df['Marks'].shift(1))
```

```python
#df.copy():It creates an separate dataframe so changes don't directly affect the original.
df = pd.DataFrame({
    'Marks': [80, 90]
})

new_df = df.copy()

print(new_df)
```

```python
#df.all(): It checks whether every value is True.
print(df.all())
```

```python
#df.empty:It returns True when the dataframe has no data.
df = pd.DataFrame()

print(df.empty)
```

```python
#df.any():It checks whether at least one value is True in a row or column.
df = pd.DataFrame({
    'A': [True, False, False],
    'B': [False, False, True]
})

print(df.any())
```

```python
#df.value_counts():It counts how many times each unique value appears.
df = pd.DataFrame({
    'City': ['Vijayawada', 'Hyderabad', 'Vijayawada', 'Vizag']
})
print(df['City'].value_counts())
```

```python
#df.unique(): Returns the unique (non-repeating) values in a column.
df = pd.DataFrame({
    'City': ['Vijayawada', 'Hyderabad', 'Vijayawada', 'Vizag']
})
print(df['City'].unique())
```

```python
#df.nunique(): Returns the count of unique values in a column.
print(df['City'].nunique())
```

```python
#df.nunique(): It returns the count of unique values in a column.
print(df.drop_duplicates())
```

```python
#df.pivot_table(): It summarizes data and displays it in a table format with rows/columns."
df = pd.DataFrame({
    'Branch': ['CSE', 'CSE', 'ECE', 'ECE'],
    'Year': [1, 2, 1, 2],
    'Marks': [80, 90, 70, 60]
})
print(df.pivot_table(values='Marks', index='Branch', columns='Year'))
```

```python
#df.reset_index()::It resets the row index so it starts from 0 again.
df = pd.DataFrame({
    'Marks': [80, 90, 70]
}, index=[5, 2, 8])
print(df.reset_index(drop=True))
```

```python
#df.sample(): It selects random rows from the dataframe.
df = pd.DataFrame({
    'Name': ['Ram', 'Sita', 'Ravi', 'Anu']
})
print(df.sample(2))
```

```python
#df.cumsum(): It gives the running (cumulative) total of values up to each row.
df = pd.DataFrame({
    'Marks': [10, 20, 30]
})
print(df['Marks'].cumsum())
```
